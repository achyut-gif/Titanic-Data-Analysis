Analysis 1 : Survival rate

```python
survival_count = ddf['survived'].sum() # no of survived people
survival_percentage = ddf['survived'].mean() * 100 # percentage of survived people
print(f"{survival_count} people survived which is {survival_percentage:.2f}%")
```

Interpretation : Only 38.13% of total passengers survived which is 498

Analysis 2: Did sex affect their survival?

```python
ddf.head()
```

```python
result = ddf.groupby('sex', as_index = False)['survived'].agg(
    total_passengers = "count",
    survivors = "sum",
    rate = "mean"
)

result['rate'] = result['rate'] * 100

print(result)
```

Interpretation : 
72.62 % of females survived whose number was 464
19.12 % of males survived whose number was 161


Analysis 3 : Did passenger class influence survival?

```python
result3 = ddf.groupby('pclass', as_index = False)['survived'].agg(
    total = 'count',
    scount = 'sum',
    srate = 'mean'
)

result3['srate'] = result3['srate'] * 100

print(result3)
```

Interpretation :
61.68 % people of class 1 survived whose number was 198
42.96 % people of class 2 survived whose number was 119
25.56 % people of class 3 survived whose number was 181


Analysis 4 : Did age have any relationship with the survivor?

```python
result4 = ddf.groupby('survived')['age'].agg(
    total = 'count',
    mean = 'mean',
    median = 'median'
)

print(result4)
```

Interpretation : 
Survivors were slightly younger in are than non-survivors. The mean age was 28.73 years for survivors compared with 29.41 years for non-survivors, while the median ages were 26.5 and 25 years respectively.



Analysis 5 : Did ticket fare relate to the survivals?

```python
result5 = ddf.groupby("survived")['fare'].agg(
    mean = 'mean',
    median = 'median'
)

print(result5)
```

Interpretation:
Survivors paid almost double the ticket fare for travelling than the non-survivors, where the average fare paid by survivor was 49.23 and by non survivor was 23.35, and the median fare was 26.0 and 10.5

3. Data Visualization

```python
fig, ax = plt.subplots(3,2)
x = [1, 2, 3]

# survival
y1 = [survival_count, len(df)-survival_count]
ax[0, 0].bar(x[0], y1[0], label = "Survived")
ax[0, 0].bar(x[1], y1[1], label = "Not Survived")
ax[0,0].set_title("Survival")
ax[0,0].set_ylabel("No of people")
ax[0,0].legend(loc = "upper left", fontsize = 8)

# survival by sex
y1 = [survival_count, len(df)-survival_count]
ax[0, 1].bar(x[0], y1[0], label = (result['sex'].iloc[0]))
ax[0, 1].bar(x[1], y1[1], label = (result['sex'].iloc[1]))
ax[0, 1].set_title("Survivors by sex")
ax[0, 1].set_ylabel("No of people")
ax[0, 1].legend(loc = "upper left", fontsize = 8)

# survival by class
y2 = [result3['scount'].iloc[0], result3['scount'].iloc[1], result3['scount'].iloc[2] ]
ax[1, 0].bar(x[0], y2[0])
ax[1, 0].bar(x[1], y2[1])
ax[1, 0].bar(x[2], y2[2])
ax[1, 0].set_title("Survivors by class")
ax[1, 0].set_ylabel("No of people")
ax[1, 0].set_xlabel("Pclass")
ax[1, 0].legend(loc = "upper left", fontsize = 8)

# survival by age
survived = ddf[ddf['survived'] == 1]['age']
not_survived = ddf[ddf['survived'] == 0]['age']

ax[1, 1].hist(survived, bins = 20, alpha = 0.6, label = "Survived")
ax[1, 1].hist(not_survived, bins = 20, alpha = 0.6, label = "Didn't survive")

ax[1, 1].set_title("Survival by age")
ax[1, 1].set_xlabel("Age")
ax[1, 1].set_ylabel("No of passengers")
ax[1, 1].legend()


plt.tight_layout()
plt.show()
```
